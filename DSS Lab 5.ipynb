```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'
```

```python
iris_df = pd.read_csv(url, header=None, names=['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'class'])
iris_df = iris_df.drop('class', axis=1)
```

```python
iris_df.isnull().sum()
```

```python
plt.scatter(iris_df['sepal_length'], iris_df['sepal_width'])
plt.title('Sepal length vs Sepal width')
plt.xlabel('Sepal length (cm)')
plt.ylabel('Sepal width (cm)')
plt.show()
```

```python
plt.scatter(iris_df['petal_length'], iris_df['petal_width'])
plt.title('Petal length vs Petal width')
plt.xlabel('Petal length (cm)')
plt.ylabel('Petal width (cm)')
plt.show()
```

```python
plt.scatter(iris_df['sepal_length'], iris_df['petal_length'])
plt.title('Sepal length vs Petal length')
plt.xlabel('Sepal length (cm)')
plt.ylabel('Petal length (cm)')
plt.show()
```
