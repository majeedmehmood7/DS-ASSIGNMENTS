```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
```

```python
df = pd.read_csv(r"student_info.csv")
df 
```

```python
plt.scatter(x= df.student_marks, y = df.study_hours)
plt.xlabel("Student_Marks")
plt.ylabel("Study_Hours")
plt.title("Students")
plt.show()
```

```python
X = df.drop("student_marks", axis = "columns")
Y = df.drop("study_hours", axis = "columns")
print("Shape of X = ",X.shape)
print("shape of y ",Y.shape)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test ,Y_train , Y_test = train_test_split(X,Y , test_size = 0.2, random_state = 51)
print("Shape of X_train = ", X_train.shape)
print("Shape of Y_train = ", Y_train.shape)
print("Shape of X_Test = ", X_test.shape)
print("Shape of Y_Test = ", Y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
Ir = LinearRegression()
Ir.fit(X_train,Y_train)
```

```python
plt.scatter(X_test,Y_test)
plt.plot(X_train,Ir.predict(X_train),color = 'r')
```

```python
Ir.intercept_
```

```python
Ir.coef_
```

```python

m = 3.91306058
c = 50.59564059
y = m * 7.10+c
y
```

```python
tmp_new = float(input("Enter the hours: "))
Ir.predict([[tmp_new]])
```

```python
y_pred = Ir.predict(X_test)
y_pred

pd.DataFrame(np.c_[X_test, Y_test, y_pred], columns = ["Study_hours","Original_Marks","predicted_Marks"] )
```
