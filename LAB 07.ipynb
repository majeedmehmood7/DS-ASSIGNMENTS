```python
from tkinter import *
from tkinter import messagebox
from tkinter.ttk import Combobox

root = Tk()
root.title("User Registration Form")


def click():
    if len(nameentry.get()) >= 1 and len(phonevalue.get()) >= 1 and len(emailentry.get()) >= 1 and var.get() >= 1 and \
            len(cityvalue.get()) >= 1 and len(statevalue.get()) >= 1:
        messagebox.showinfo("Success", "Registration Completed Successfully!")
    else:
        messagebox.showinfo("Error", "Registration Unsuccessfull!")


root.geometry("644x344")

Label(root, text="Registration Form", font="comicsansms 18 bold", pady=15).grid(row=0, column=1)

Label(root, text="Name *").grid(row=1, column=0, padx=20, pady=5, sticky='w')
Label(root, text="Contact *").grid(row=2, column=0, padx=20, pady=5, sticky='w')
Label(root, text="Email *").grid(row=3, column=0, padx=20, pady=5, sticky='w')
Label(root, text="Gender *").grid(row=4, column=0, padx=20, pady=5, sticky='w')
Label(root, text="City *").grid(row=5, column=0, padx=20, pady=5, sticky='w')
Label(root, text="State *").grid(row=6, column=0, padx=20, pady=5, sticky='w')

namevalue = StringVar()
phonevalue = StringVar()
emailvalue = StringVar()
cityvalue = Combobox(root)
statevalue = Combobox(root)
var = IntVar()

nameentry = Entry(root, textvariable=namevalue)
phoneentry = Entry(root, textvariable=phonevalue)
emailentry = Entry(root, textvariable=emailvalue)
genderentry1 = Radiobutton(root, variable=var, text="Male", value=1)
genderentry2 = Radiobutton(root, variable=var, text="Female", value=2)
cityvalue['values'] = ("Karachi", "Islamabad", "Lahore", "Peshawar", "Hyderabad", "Quetta")
statevalue['values'] = ("Sindh", "Punjab", "Balochistan", "KPK", "Gilgit Baltistan")

nameentry.grid(row=1, column=1)
phoneentry.grid(row=2, column=1)
emailentry.grid(row=3, column=1)
genderentry1.grid(row=4, column=1)
genderentry2.grid(row=4, column=2)
cityvalue.grid(row=5, column=1)
statevalue.grid(row=6, column=1)

Button(text="Register", bg="orange", padx=10, command=click).grid(row=7, column=1, pady=15)

root.mainloop()
```

```python
from tkinter import *

window = Tk()

window.geometry("512x455")

window.resizable(False, False)

window.title("Calcualtor")


def btn_click(item):
    global expression
    expression = expression + str(item)
    input_text.set(expression)


def btn_clear():
    global expression
    expression = ""
    input_text.set("")


def btn_equal():
    global expression
    result = str(eval(expression))
    input_text.set(result)
    expression = ""


expression = ""

input_text = StringVar()

input_frame = Frame(window, width=312, height=50, bd=0, highlightbackground="black", highlightcolor="black",
                    highlightthickness=1)

input_frame.pack(side=TOP)

input_field = Entry(input_frame, font=('arial', 18, 'bold'), textvariable=input_text, width=50, bg="#eee", bd=0,
                    justify=RIGHT)

input_field.grid(row=0, column=0)

input_field.pack(ipady=10)

btns_frame = Frame(window, width=312, height=272.5, bg="grey")

btns_frame.pack()

Button(btns_frame, text="C", font="lucida 14 bold", fg="black", width=32, height=3, bd=0, bg="#eee", cursor="hand2",
       command=lambda: btn_clear()).grid(row=0, column=0, columnspan=3, padx=1, pady=1)

Button(btns_frame, text="/", font="lucida 14 bold", fg="black", width=10, height=3, bd=0, bg="#eee", cursor="hand2",
       command=lambda: btn_click("/")).grid(row=0, column=3, padx=1, pady=1)

Button(btns_frame, text="7", font="lucida 14 bold", fg="black", width=10, height=3, bd=0, bg="#fff", cursor="hand2",
       command=lambda: btn_click(7)).grid(row=1, column=0, padx=1, pady=1)

Button(btns_frame, text="8", font="lucida 14 bold", fg="black", width=10, height=3, bd=0, bg="#fff", cursor="hand2",
       command=lambda: btn_click(8)).grid(row=1, column=1, padx=1, pady=1)

Button(btns_frame, text="9", font="lucida 14 bold", fg="black", width=10, height=3, bd=0, bg="#fff", cursor="hand2",
       command=lambda: btn_click(9)).grid(row=1, column=2, padx=1, pady=1)

Button(btns_frame, text="*", font="lucida 14 bold", fg="black", width=10, height=3, bd=0, bg="#eee", cursor="hand2",
       command=lambda: btn_click("*")).grid(row=1, column=3, padx=1, pady=1)

Button(btns_frame, text="4", font="lucida 14 bold", fg="black", width=10, height=3, bd=0, bg="#fff", cursor="hand2",
       command=lambda: btn_click(4)).grid(row=2, column=0, padx=1, pady=1)

Button(btns_frame, text="5", font="lucida 14 bold", fg="black", width=10, height=3, bd=0, bg="#fff", cursor="hand2",
       command=lambda: btn_click(5)).grid(row=2, column=1, padx=1, pady=1)

Button(btns_frame, text="6", font="lucida 14 bold", fg="black", width=10, height=3, bd=0, bg="#fff", cursor="hand2",
       command=lambda: btn_click(6)).grid(row=2, column=2, padx=1, pady=1)

Button(btns_frame, text="-", font="lucida 14 bold", fg="black", width=10, height=3, bd=0, bg="#eee", cursor="hand2",
       command=lambda: btn_click("-")).grid(row=2, column=3, padx=1, pady=1)

Button(btns_frame, text="1", font="lucida 14 bold", fg="black", width=10, height=3, bd=0, bg="#fff", cursor="hand2",
       command=lambda: btn_click(1)).grid(row=3, column=0, padx=1, pady=1)

Button(btns_frame, text="2", font="lucida 14 bold", fg="black", width=10, height=3, bd=0, bg="#fff", cursor="hand2",
       command=lambda: btn_click(2)).grid(row=3, column=1, padx=1, pady=1)

Button(btns_frame, text="3", font="lucida 14 bold", fg="black", width=10, height=3, bd=0, bg="#fff", cursor="hand2",
       command=lambda: btn_click(3)).grid(row=3, column=2, padx=1, pady=1)

Button(btns_frame, text="+", font="lucida 14 bold", fg="black", width=10, height=3, bd=0, bg="#eee", cursor="hand2",
       command=lambda: btn_click("+")).grid(row=3, column=3, padx=1, pady=1)

Button(btns_frame, text="0", font="lucida 14 bold", fg="black", width=21, height=3, bd=0, bg="#fff", cursor="hand2",
       command=lambda: btn_click(0)).grid(row=4, column=0, columnspan=2, padx=1, pady=1)

Button(btns_frame, text=".", font="lucida 14 bold", fg="black", width=10, height=3, bd=0, bg="#eee", cursor="hand2",
       command=lambda: btn_click(".")).grid(row=4, column=2, padx=1, pady=1)

Button(btns_frame, text="=", font="lucida 14 bold", fg="black", width=10, height=3, bd=0, bg="#eee", cursor="hand2",
       command=lambda: btn_equal()).grid(row=4, column=3, padx=1, pady=1)

window.mainloop()
```
